### 30. Clipboard workflow: display table, columns, select columns, add a column, then delete it.

```python
# Copy a table with at least four columns (including headers) before running this cell.
lab_df = pd.read_clipboard()

display(lab_df)
display(lab_df.columns)

# Retrieve a single-word column: replace 'Name' if your copied table uses another heading.
# Retrieve a column containing a space: replace 'Total Marks' if needed.
# The try/except blocks help the cell run even if your table has different headings.
if 'Name' in lab_df.columns:
    display(lab_df['Name'])
else:
    display(lab_df.iloc[:, 0])

space_columns = [col for col in lab_df.columns if ' ' in str(col)]
if space_columns:
    display(lab_df[space_columns[0]])
else:
    print('No column with a space in its name was found.')

selected_columns = list(lab_df.columns[:3])
small_lab_df = lab_df[selected_columns]
display(small_lab_df)

lab_df['NewColumn'] = 'Practice'
display(lab_df)

del lab_df['NewColumn']
display(lab_df)
```
