```python
#Q27
d = {'Red': 12, 'Blue': 18, 'Green': 24}
labels = ['Green', 'Red', 'Yellow', 'Blue']

result = pd.Series(d, index=labels)

# (a) Display the Series
print("(a) Result:")
print(result)

# (b) Check the missing value
print("\n(b) pd.isnull(result):")
print(pd.isnull(result))

# (c) Explain why the value is missing
print("\n(c) Reason:")
print("Yellow is missing because it is not present in dictionary d.")

# (d) Labels retaining numeric values
print("\n(d) Labels with actual numeric values:")
print(result.dropna())
```

```python
#Q28
s = pd.Series([90, 75, 82], index=['Ravi', 'Aman', 'Neha'])

d = s.to_dict()
s2 = pd.Series(d)

print("(a) Original Series:")
print(s)

print("\n(b) Dictionary:")
print(d)

print("\n(c) Series s2:")
print(s2)

print("\n(d) Index of s2:")
print(s2.index.tolist())

print("\n(e) Values of s2:")
print(s2.values.tolist())

print("\n(f) Do s and s2 represent the same labelled data?")
print(s.equals(s2))
```

```python
#Q29

sales = pd.Series(
    [450, 700, 550, 300],
    index=['Laptop', 'Mobile', 'Tablet', 'Keyboard']
)

# (a) Display values and index
print("(a) Sales Series:")
print(sales)

print("\nIndex:")
print(sales.index.tolist())

# (b) Select one label
print("\n(b) Sales of Mobile:")
print(sales['Mobile'])

# (c) Filter sales above 500
print("\n(c) Sales above 500:")
print(sales[sales > 500])

# (d) Check whether a label exists
print("\n(d) Does Tablet exist?")
print('Tablet' in sales.index)

# (e) Convert Series to dictionary
print("\n(e) Series as dictionary:")
print(sales.to_dict())
```

```python
#Q30
data = {'A': 15, 'B': 25, 'C': 35}
s = pd.Series(data, index=['C', 'A', 'D', 'B'])

# (a) Display index order
print("(a) Index order:")
print(s.index.tolist())

# (b) Find the label containing NaN
print("\n(b) Label containing NaN:")
print(s[s.isnull()].index.tolist())

# (c) Boolean filter: values greater than 20
print("\n(c) Values where s > 20:")
print(s[s > 20])

# (d) Check non-null values for each label
print("\n(d) pd.notnull(s):")
print(pd.notnull(s))

# (e) Assign Series name and index name
s.name = 'Observations'
s.index.name = 'Category'

print("\n(e) Final Series:")
print(s)

print("\nSeries name:", s.name)
print("Index name:", s.index.name)
```
