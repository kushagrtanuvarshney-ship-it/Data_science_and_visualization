**Section 2 — Array Arithmetic**

```python
#Q2.1
a = np.array([10,20,30])
b = np.array([2,4,5])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

**Section 3 — Indexing & Slicing**

```python
#Q3.1
a = np.arange(21)
print(a[:5])
print(a[-5:])
print(a[5:16])
print(a[::2])
```

```python
#Q3.2
a = np.arange(10)
a[:5] = 100
print(a)
```

```python
#Q3.3
a = np.array([1,2,3,4,5])
b = a[1:4]
c = a[1:4].copy()
b[0] = 100
c[0] = 200
print(a)
print(b)
print(c)
```

```python
#Q3.4
a = np.array([1,2,3,4,5])
print(a[::-1])
```

**Section 8 — np.where()**

```python
#Q8.1
marks = np.array([35,72,48,90,55,28])
print(np.where(marks<40,0,marks))
```

```python
#Q8.2
income = np.array([30000,55000,45000,70000])
print(np.where(income>=50000,"High","Low"))
```

**Section 9 — Statistical Operations**

Q11.1 — Student Marks Analysis

```python
#Q11.1
marks = np.array([[60,70,80],[40,50,60],[30,40,35],[80,90,85],[55,65,60],
                  [45,50,55],[70,75,80],[35,30,40],[90,85,95],[50,55,60]])
total = marks.sum(axis=1)
average = marks.mean(axis=1)
print(total)
print(average)
print(np.where(average>=50)[0])
```
