```python
#36

matrix = np.arange(21, 46).reshape(5, 5)

central = matrix[1:4, 1:4]
transposed = central.T
product = central @ transposed

row_sums = np.sum(product, axis=1)
maximum = np.max(product)
positions = np.argwhere(product == maximum)

print("Original matrix:\n", matrix)
print("Central matrix:\n", central)
print("Transpose:\n", transposed)
print("Product:\n", product)
print("Row sums:", row_sums)
print("Maximum:", maximum)
print("Positions (zero-based):\n", positions)
```

```python
#37

quantity = np.array([
    [10, 20, 15, 8],
    [12, 18, 20, 10],
    [25, 15, 12, 14],
    [20, 25, 18, 12],
    [18, 22, 16, 15],
    [22, 20, 25, 18]
])

prices = np.array([150, 240, 90, 320])

daily_revenue = quantity @ prices
product_quantity = np.sum(quantity, axis=0)
product_revenue = product_quantity * prices

highest_day = np.argmax(daily_revenue) + 1
best_product = np.argmax(product_quantity) + 1

new_prices = np.where(product_quantity > 100, prices * 0.90, prices)
new_daily_revenue = quantity @ new_prices

print("Daily revenue:", daily_revenue)
print("Revenue per product:", product_revenue)
print("Total revenue:", np.sum(daily_revenue))
print("Highest-revenue day:", highest_day)
print("Quantity per product:", product_quantity)
print("Best-selling product:", best_product)
print("New prices:", new_prices)
print("Revenue after discount:", np.sum(new_daily_revenue))
```

```python
#38

np.random.seed(42)

data = np.random.normal(50, 12, 100)

outside = np.sum((data < 30) | (data > 70))

cleaned = np.where(data < 30, 30,
                   np.where(data > 70, 70, data))

print("Values outside 30-70:", outside)

print("Original mean:", np.mean(data))
print("Cleaned mean:", np.mean(cleaned))

print("Original variance:", np.var(data))
print("Cleaned variance:", np.var(cleaned))

print("Original standard deviation:", np.std(data))
print("Cleaned standard deviation:", np.std(cleaned))

print("Original minimum and maximum:", np.min(data), np.max(data))
print("Cleaned minimum and maximum:", np.min(cleaned), np.max(cleaned))
```

```python
#39

branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

result = branch_a + branch_b + branch_c

common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
common_earnings = result.loc[common]

zero_result = (
    branch_a.add(branch_b, fill_value=0)
            .add(branch_c, fill_value=0)
)

print("Ordinary addition:")
print(result)

print("Quarters common to all branches:")
print(common_earnings)

print("Addition treating missing records as zero:")
print(zero_result)
```

```python
#40

scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']

s = pd.Series(scores, index=required)

missing = s[s.isnull()]
valid = s[s >= 70]

data = valid.to_dict()
reconstructed = pd.Series(data)

print("Original Series:")
print(s)

print("Missing student records:")
print(missing)

print("Valid scores >= 70:")
print(valid)

print("Dictionary:", data)
print("Reconstructed Series:")
print(reconstructed)

print("Values match:", valid.equals(reconstructed))
```
