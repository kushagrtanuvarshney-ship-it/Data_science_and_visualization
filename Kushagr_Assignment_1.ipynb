```python
#Q3.2
TRUE
FALSE

a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
print(a is b)
```

**Topic 6: Conditions — if / elif / else, modulo (%)**

```python
#Q6.1

for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
#Q6.2

for i in range(1, 16):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
#Q6.3

#OUTPUT - 0,11,-8,3,14

for i in range(0, 5):
    if i % 3 == 0:
        print(i)
    elif i % 3 == 1:
        print(i + 10)
    else:
        print(i - 10)
```
